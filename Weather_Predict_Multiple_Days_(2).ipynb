```python
import requests

# 東京の緯度・経度
latitude = 35.6895
longitude = 139.6917

# Open-Meteo API（dailyデータ）
url = (
    f"https://api.open-meteo.com/v1/forecast"
    f"?latitude={latitude}&longitude={longitude}"
    f"&daily=temperature_2m_mean,precipitation_sum,windspeed_10m_mean,relative_humidity_2m_mean"
    f"&timezone=Asia%2FTokyo"
)

response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    daily = data["daily"]

    print("📅 7日間の天気予報（東京）")
    print("=" * 40)
    for i in range(len(daily["time"])):
        date = daily["time"][i]
        temp = daily["temperature_2m_mean"][i]
        rain = daily["precipitation_sum"][i]
        wind = daily["windspeed_10m_mean"][i]
        humidity = daily["relative_humidity_2m_mean"][i]
        print(f"{date} ▶ 平均気温: {temp}℃, 降水量: {rain}mm, 風速: {wind}m/s, 湿度: {humidity}%")
else:
    print("データの取得に失敗しました")
```

```python
beer_columns = [
    "ペールエール(本)", "ペールエール(円)",
    "ラガー(本)", "ラガー(円)",
    "IPA(本)", "IPA(円)",
    "ホワイトビール(本)", "ホワイトビール(円)",
    "黒ビール(本)", "黒ビール(円)",
    "フルーツビール(本)", "フルーツビール(円)"
]

df_merge[beer_columns] = df_merge[beer_columns].fillna(0)

weather_columns = [
    "平均気温(℃)",
    "降水量の合計(mm)",
    "合計全天日射量(MJ/㎡)",
    "最深積雪(cm)",
    "平均風速(m/s)",
    "平均蒸気圧(hPa)",
    "平均現地気圧(hPa)",
    "平均湿度(％)"
]
for col in weather_columns:
    median = df_merge[col].median()
    df_merge[col].fillna(median, inplace=True)
df_merge["天気概況(昼：06時～18時)"] = df_merge["天気概況(昼：06時～18時)"].fillna("不明")
```

```python
from sklearn.preprocessing import LabelEncoder

counts = df_merge["天気概況(昼：06時～18時)"].value_counts()
rare_labels = counts[counts <= 2].index
df_merge["天気概況(昼：06時～18時)"] = df_merge["天気概況(昼：06時～18時)"].replace(rare_labels, "その他")

le = LabelEncoder()
df_merge["天気概況コード"] = le.fit_transform(df_merge["天気概況(昼：06時～18時)"])

for i, label in enumerate(le.classes_):
    print(f"{i} → {label}")
```

```python
weekday_map = {"月": 0, "火": 1, "水": 2, "木": 3, "金": 4, "土": 5, "日": 6}
df_merge["曜日番号"] = df_merge["曜日"].map(weekday_map)
```

```python
from datetime import datetime
daily = data["daily"]
dates = daily["time"]
temps = daily["temperature_2m_mean"]
rains = daily["precipitation_sum"]
winds = daily["windspeed_10m_mean"]
humidities = daily["relative_humidity_2m_mean"]


predictions = []

for date, temp, rain, wind, humidity in zip(dates, temps, rains, winds, humidities):
    weekday_num = datetime.strptime(date,"%Y-%m-%d").weekday()
    features = [[temp, rain, wind, humidity,weekday_num]]
    model = DecisionTreeRegressor(random_state=42)
    model.fit(X,y)
    pred = model.predict(features)[0]
    predictions.append([date, temp, rain, wind, humidity] + list(pred))

columns = ["日付", "平均気温（℃）", "降水量（mm）", "風速（m/s）", "湿度（％）",
           "ペールエール（本）", "ラガー（本）", "IPA（本）", "ホワイトビール（本）", "黒ビール（本）", "フルーツビール（本）"]

df_pred = pd.DataFrame(predictions, columns=columns)


```

```python
df_pred["日付"] = pd.to_datetime(df_pred["日付"])

weekday_map = {0: "月", 1: "火", 2: "水", 3: "木", 4: "金", 5: "土", 6: "日"}
df_pred["曜日"] = df_pred["日付"].dt.weekday.map(weekday_map)

first_col = df_pred.pop("曜日")
df_pred.insert(1, "曜日", first_col)

print("📦 未来7日間のビール売上予測：")
display(df_pred)
```

```python
df_pred["日付"] = pd.to_datetime(df_pred["日付"])

monday_group = df_pred[df_pred["日付"].dt.weekday.isin([1, 2, 3])]
monday_sum = monday_group.filter(like="（本）").sum().to_frame(name="月曜用の出荷集計").T

thursday_group = df_pred[df_pred["日付"].dt.weekday.isin([4, 5, 0])]
thursday_sum = thursday_group.filter(like="（本）").sum().to_frame(name="木曜用の出荷集計").T

shipment_summary = pd.concat([monday_sum, thursday_sum])

print("🍻 発注用ビール出荷集計:")
print(shipment_summary)
```
